### Using the Breast Cancer Wisconsin (Diagnostic) dataset, write a small program (Python?) that implements logistic regression from scratch to predict whether a tumor is benign or malignant. Your program should select two or three numerical features from the dataset (don't select all), initialize the model weights and bias randomly, calculate the predicted probability using the sigmoid function, and use gradient descent to update the weights and bias over multiple training iterations. After training, use the model to classify the test data and report the classification accuracy. Do not use a machine-learning library or an existing logistic regression implementation, etc.

```python
import random

weight = [random.uniform(-0.1, 0.1) for _ in range(3)]
bias = random.uniform(-0.1, 0.1)

print(f"Randomized Weights: [{weight[0]}, {weight[1]}, {weight[2]}]")
print("Randomized Bias:", bias)
```

```python
import math
import csv

def predict(weight, inputData, bias):
    value = 0.0
    for i in range(len(weight)):
        value += weight[i] * inputData[i]
    z = value + bias
    return sigmoid(z)


def sigmoid(z):
    z = max(-500, min(500, z))
    return 1 / (1 + math.exp(-z))

def gradientDescent(weight, inputData, diagnosis, bias, learningRate, trainingSteps):
    n = len(inputData)

    for i in range(trainingSteps):
        dWeight = [0.0] * len(weight)
        dBias = 0.0

        for k in range(n):
            probability = predict(weight, inputData[k], bias)
            error = diagnosis[k] - probability
            dz = -2 * error * probability * (1 - probability)
            for j in range(len(weight)):
                dWeight[j] += inputData[k][j] * dz
            dBias += dz

        for j in range(len(dWeight)):
            dWeight[j] /= n
            weight[j] -= learningRate * dWeight[j]
        
        dBias /= n
        bias -= learningRate * dBias
    
    return weight, bias

def classificationAccuracy(weight, inputData, diagnosis, bias):
    correct = 0
    for i in range(len(inputData)):
        probability = predict(weight, inputData[i], bias)
        p_class = 1 if probability >= 0.5 else 0
        if(p_class == diagnosis[i]):
            correct += 1
    
    accuracy = correct / len(diagnosis) * 100
    return accuracy
    

#----------------------------------------------#

inputData = []
diagnosis = []
with open("wdbc.data", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        # row[1]: malignant = 1, benign = 0
        if(row[1] == "M"):
            diagnosis.append(1)
        else:
            diagnosis.append(0)
        
        # row[2]: radius;     row[3]: texture;      row[4]: perimeter;
        inputData.append([float(row[2]), float(row[3]), float(row[4])])

learningRate = 0.01
trainingSteps = 10000

print("Number of samples:", len(inputData))
print("Number of features:", len(inputData[0]))
print("Learning Rate:", learningRate)
print("Training Steps:", trainingSteps, "\n")

accuracy = classificationAccuracy(weight, inputData, diagnosis, bias)
print(f"Current weight before training: [{weight[0]}, {weight[1]}, {weight[2]}]")
print("Current bias before training:", bias, "\n")
print("Accuracy before training:", accuracy)

weight, bias = gradientDescent(weight, inputData, diagnosis, bias, learningRate, trainingSteps)
accuracy = classificationAccuracy(weight, inputData, diagnosis, bias)
print(f"New weight after training: [{weight[0]}, {weight[1]}, {weight[2]}]")
print("New bias after training:", bias)
print("Accuracy after training:", accuracy)
```
