### HW1: Using the SMS Spam Collection dataset, write a small program (Python?) that implements a Naive Bayes spam classifier from scratch. Your program should calculate the prior probabilities of spam and non-spam messages, estimate the probability of selected words appearing in each class, and use Bayesâ€™ rule to classify new messages as spam or non-spam. Assume that the words in a message are conditionally independent given the class. Test your classifier on a small set of messages and print the predicted class and probability for each message. Do not use a machine-learning library or an existing Naive Bayes implementation; implement the probability calculations yourself.

```python
def calculate(dataset):
    spam = 0
    ham = 0
    for line in dataset:
        classifier, sentence = line.split(maxsplit=1)
        if(classifier == "spam"):
            spam+=1
        elif(classifier == "ham"):
            ham+=1

    spam_prob = spam / len(dataset)
    ham_prob = ham / len(dataset)
    return spam, ham, spam_prob, ham_prob

def estimateWords(dataset, word_counts):
    for line in dataset:
        classifier, sentence = line.split(maxsplit=1)
        words = sentence.split()
        if(classifier == "spam"):
            for i in words:
                word_counts["spam"][i] = word_counts["spam"].get(i, 0) + 1
        elif(classifier == "ham"):
            for i in words:
                word_counts["ham"][i] = word_counts["ham"].get(i, 0) + 1

def predict(sentence, ham_prob, spam_prob, word_counts, total_ham_words, total_spam_words, vocabulary_size):
    ham_score = ham_prob
    spam_score = spam_prob
    user_words = sentence.split()

    for word in user_words:
        ham_s = P(word, word_counts["ham"], total_ham_words, vocabulary_size)
        spam_s = P(word, word_counts["spam"], total_spam_words, vocabulary_size)

        print(word)
        print("   P(word|ham):", ham_s)
        print("   P(word|spam):", spam_s,"\n")

        ham_score *= ham_s
        spam_score *= spam_s

    total_score = ham_score + spam_score
    h_prob = ham_score / total_score
    s_prob = spam_score / total_score
    if(h_prob < s_prob):
        return f"inputted message is spam, ham probability: {h_prob}, spam probability: {s_prob} \n\n\n"
    else:
        return f"inputted message is ham, ham probability: {h_prob}, spam probability: {s_prob} \n\n\n"

def P(word, word_counts, total_words, vocabulary_size):
    count = word_counts.get(word, 0)
    return (count + 1) / (total_words + vocabulary_size)

#---------------------------------------------------#
dataset = []
word_counts = {
    "spam": {},
    "ham": {}
}

with open("SMSSpamCollection.txt") as f:
    for line in f:
        dataset.append(line.strip().lower())

spam, ham, spam_prob, ham_prob = calculate(dataset)
estimateWords(dataset, word_counts)
total_ham_words = sum(word_counts["ham"].values())
total_spam_words = sum(word_counts["spam"].values())
vocabulary_size = len(set(word_counts["spam"] | word_counts["ham"]))

#---------------------------------------------------#
print("Dataset size: ", len(dataset))
print(f"Total spam: {spam}, Total ham: {ham}")
print(f"Spam probability: {spam_prob}, Ham probability: {ham_prob}\n")

print("Number of unique words in spam:", len(word_counts["spam"]))
print("Number of unique words in ham:", len(word_counts["ham"]),"\n")


#---------------------------------------------------#
testSentences = ["Hey man, did you receive the message I've sent yesterday?",
                 "Claim your free prize today! Win free money!",
                 "What homework is due next week?"]

print("Testing some sentences on Naive Bayes")
print("-------------------------------------")
for i in testSentences:
    print(i)
    print(predict(i.lower(), ham_prob, spam_prob, word_counts, total_ham_words, total_spam_words, vocabulary_size))

```

### Testing for User input

```python
while(True):
    print("Enter a new sentence (type !q to terminate): ")
    sentence = input()
    print(sentence)
    if(sentence.lower() == "!q"):
        break
    print(predict(sentence.lower(), ham_prob, spam_prob, word_counts, total_ham_words, total_spam_words, vocabulary_size),"\n")

print("Program terminated")
```
