### HW1: Using the SMS Spam Collection dataset, write a small program (Python?) that implements a Naive Bayes spam classifier from scratch. Your program should calculate the prior probabilities of spam and non-spam messages, estimate the probability of selected words appearing in each class, and use Bayesâ€™ rule to classify new messages as spam or non-spam. Assume that the words in a message are conditionally independent given the class. Test your classifier on a small set of messages and print the predicted class and probability for each message. Do not use a machine-learning library or an existing Naive Bayes implementation; implement the probability calculations yourself.

```python
def calculate(dataset):
    spam = sum(1 for i in dataset if "spam" in i)
    ham = sum(1 for i in dataset if "ham" in i)
    spam_prob = spam / len(dataset)
    ham_prob = ham / len(dataset)
    return spam, ham, spam_prob, ham_prob

def estimateWords(dataset, word_counts):
    for line in dataset:
        classifier, sentence = line.split(maxsplit=1)
        words = sentence.split()
        if(classifier == "spam"):
            for i in words:
                word_counts["spam"][i] = word_counts["spam"].get(i, 0) + 1
        elif(classifier == "ham"):
            for i in words:
                word_counts["ham"][i] = word_counts["ham"].get(i, 0) + 1

def predict(sentence, ham_prob, spam_prob, word_counts, vocabulary_size):
    ham_score = ham_prob
    spam_score = spam_prob
    user_words = sentence.split()

    for word in user_words:
        ham_s = P(word, word_counts["ham"], vocabulary_size)
        spam_s = P(word, word_counts["spam"], vocabulary_size)

        print(word)
        print("   ham probability:", ham_s)
        print("   spam probability:", spam_s)

        ham_score *= ham_s
        spam_score *= spam_s

    if(ham_score < spam_score):
        return f"inputted message is spam, spam score: {spam_score}, ham score: {ham_score}"
    else:
        return f"inputted message is ham, ham score: {ham_score}, spam score: {spam_score}"

def P(word, word_counts, vocabulary_size):
    count = word_counts.get(word, 0)
    total_words = sum(word_counts.values())
    return (count + 1) / (total_words + vocabulary_size)

#---------------------------------------------------#
dataset = []
word_counts = {
    "spam": {},
    "ham": {}
}

with open("SMSSpamCollection.txt") as f:
    for line in f:
        dataset.append(line.strip().lower())

spam, ham, spam_prob, ham_prob = calculate(dataset)
estimateWords(dataset, word_counts)
vocabulary_size = len(set(word_counts["spam"] | word_counts["ham"]))

#---------------------------------------------------#
print("dataset size: ", len(dataset))
print(f"total spam: {spam}, total ham: {ham}")
print(f"spam probability: {spam_prob}, ham probability: {ham_prob}")

print("number of words in spam:", len(word_counts["spam"]))
print("number of words in ham:", len(word_counts["ham"]))


#---------------------------------------------------#
while(True):
    print("Enter a new sentence (type !exit to terminate): ")
    sentence = input()
    print(sentence)
    if(sentence.lower() == "!exit"):
        break
    print(predict(sentence.lower(), ham_prob, spam_prob, word_counts, vocabulary_size))

```
