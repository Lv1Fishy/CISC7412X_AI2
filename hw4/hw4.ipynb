### Rewrite HW3 in PyTorch, use PyTorch's automatic differentiation and train the network using the Adam optimizer. Initialize the network weights using Xavier initialization. Train the network until it correctly reproduces all eight inputs with no errors. Tip: last layer should be softmax.

```python
import torch
import torch.nn as nn

torch.manual_seed(0)
binary_8 = torch.eye(8).flip(0)
labels = binary_8.argmax(dim=1)

hidden_layer = nn.Sequential(nn.Linear(8,3), nn.Sigmoid())
output_layer = nn.Sequential(nn.Linear(3,8), nn.LogSoftmax(dim=1))
model = nn.Sequential(hidden_layer, output_layer)
loss_function = nn.NLLLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

for layer in (hidden_layer[0], output_layer[0]):
    nn.init.xavier_uniform_(layer.weight)
    nn.init.zeros_(layer.bias)


with open("hw4_output.txt", "w", encoding="utf-8") as f:
    print("Training Autoencoder")
    f.write("Training Autoencoder\n")
    trainingSteps = 2000
    for step in range(trainingSteps):
        optimizer.zero_grad()
        loss = loss_function(model(binary_8), labels)
        loss.backward()
        optimizer.step()

        with torch.no_grad():
            predict = model(binary_8).argmax(dim=1)
            correct = (predict == labels).sum()
            codes = (hidden_layer(binary_8) >= 0.5).int()
            unique = len({tuple(c.tolist()) for c in codes})
            if(correct == 8 and unique == 8):
                print(f"All eight inputs and unique codes are correct at step {step + 1}")
                f.write(f"All eight inputs matches training output at step {step + 1}\n")
                break

        if((step + 1) % (trainingSteps//100) == 0):
            print(f"Training Step {step + 1}: | Loss: {loss.item()}")
            f.write(f"Training Step {step + 1}: | Loss: {loss.item()}\n")
```
